```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression

# 1. Bikin dataset buatan teks bahasa Indonesia
data = {
    'teks': [
        'kamu hebat banget bro semangat ya',
        'woi lah belajar ai seru banget',
        'dasar bodoh gitu aja ga bisa',
        'aplikasi ini sampah banget rusak terus',
        'terima kasih banyak atas bantuannya',
        'lo jelek banget gausah sok tau deh',
        'kamu kok bodoh banget sih',
        'eh bro jangan bodoh lah',
        'kamu emang anak hebat dan pinter bro'
    ],
    'label': [0, 0, 1, 1, 0, 1, 1, 1, 0] # 0 = Positif/Aman, 1 = Toksik/Kasar
}

df = pd.DataFrame(data)
print(df)
```

```python
# 2. Ubah teks menjadi angka
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(df['teks'])
y = df['label']

# 3. Buat dan latih model AI (Logistic Regression)
model = LogisticRegression()
model.fit(X, y)
print("Model AI Berhasil Dilatih, Bro! 😎")
```

```python
# 4. Coba masukkan kalimat baru sesukamu di sini
kalimat_baru = ["kamu bodoh banget sih bro"]

X_baru = vectorizer.transform(kalimat_baru)
prediksi = model.predict(X_baru)

if prediksi[0] == 1:
    print(f"Hasil: '{kalimat_baru[0]}' -> 🚨 TOKSIK / KASAR!")
else:
    print(f"Hasil: '{kalimat_baru[0]}' -> ✅ AMAN / POSITIF")
```
