```python
# Cell 1: Definisikan fungsi Few-Shot Prompt Builder
def buat_prompt_few_shot(kalimat_baru):
    # Di sini kita kasih "Informasi Tambahan / Contoh" (Few-Shot)
    prompt = """
Kamu adalah AI penerjemah bahasa gaul anak kosan Yogyakarta menjadi Bahasa Indonesia baku.
Ikuti pola contoh di bawah ini secara saklek!

Contoh 1:
Input: "Bro, piye iki tugas alproku error terus, mumet aku."
Output: "Teman, bagaimana ini tugas algoritma pemrogramanku mengalami kesalahan terus, saya pusing."

Contoh 2:
Input: "Aku mau jajan sego kucing dhisik nang angkringan mburung kos."
Output: "Saya ingin membeli nasi kucing terlebih dahulu di angkringan belakang tempat kos."

Contoh 3:
Input: "Labil banget aku pusing mikirin kodingan virus siang ini."
Output: "Saya merasa sangat bimbang dan pusing memikirkan logika pemrograman virus siang ini."

Sekarang giliranmu, terjemahkan kalimat di bawah ini menggunakan pola yang sama!
Input: "{kalimat}"
Output:""".format(kalimat=kalimat_baru)
    
    return prompt

# Coba kita tes hasil rakitan prompt-nya
kalimat_test = "Woi lah, piye iki pandas-ku ra iso jalan, tulung brok."
prompt_final = buat_prompt_few_shot(kalimat_test)
print(prompt_final)
```

```python
# Cell 2: Simulasi Balasan Model LLM berdasarkan Prompt Few-Shot
def simulasi_llm_respond(prompt_masukan):
    # Di dunia nyata, variabel prompt_masukan ini ditembak ke API OpenAI/Gemini
    # Di sini kita bikin simulasi kecerdasan AI buatan kita sendiri dulu
    
    if "pandas-ku ra iso jalan" in prompt_masukan:
        return '"Waduh, bagaimana ini library pandas saya tidak bisa berjalan, mohon bantuannya teman."'
    else:
        return '"Maaf, AI belum mengenali pola kalimat gaul ini."'

# Mari kita jalankan simulasinya
hasil_ai = simulasi_llm_respond(prompt_final)

print("=== HASIL RESPONS AI (FEW-SHOT MODEL) ===")
print(hasil_ai)
```

```python
# Cell 3: Menembak Prompt Few-Shot ke Google Gemini Asli!
import os
from dotenv import load_dotenv 
from google import genai

# 1. Perintahkan Python untuk membaca file .env tersembunyi
load_dotenv()

# 2. Ambil kuncinya dari memori sistem (tanpa menulis string-nya terang-terangan)
apiKey = os.getenv("GEMINI_API_KEY")

# 3. Sambungkan pipa ke server Google pakai API Key kamu
client = genai.Client(api_key=apiKey)

print("Mengirim prompt Few-Shot secara aman ke Google Gemini...")

# 4. Kirim variabel 'prompt_final' (rakitan Few-Shot dari Cell 1 tadi)
response = client.models.generate_content(
    model='gemini-2.5-flash', # Memakai model Gemini yang super cepat
    contents=prompt_final,
)

print("\n=== HASIL RESPONS GEMINI ASLI ===")
print(response.text)
```
